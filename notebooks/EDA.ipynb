# Key Business Insights

## 1. Overall Churn Distribution

The dataset contains 7,043 customers.

- No Churn: 73.46%

This indicates that customer churn is a significant business problem, while the target variable also shows class imbalance that should be considered during model training.

## 2. Gender Has Minimal Difference in Churn

The churn rates are:

- Female: 26.92%
- Male: 26.16%

The difference is very small, suggesting that gender does not show a strong association with churn in this dataset.

---

## 3. Senior Citizens Show Higher Churn

Churn rates:

- Non-senior citizens: 23.61%

Senior citizens have a substantially higher observed churn rate in this dataset.

This indicates that age-related customer characteristics may be useful for churn prediction and retention analysis.

---

## 4. Customers Without Partners Have Higher Churn

Churn rates:

- No Partner: 32.96%
- Partner: 19.66%

Customers without a partner show a higher observed churn rate.

---

## 5. Customers Without Dependents Have Higher Churn

Churn rates:

- No Dependents: 31.28%
- Dependents: 15.45%
  
This indicates a noticeable association between dependent status and customer churn.

---

## 6. Contract Type Is Strongly Associated with Churn

Observed churn rates:

- Month-to-month: 42.71%
- One year: 11.27%
- Two year: 2.83%

Customers on month-to-month contracts show a substantially higher churn rate than customers on longer-term contracts.

Contract type is therefore an important feature for the predictive model.

---

## 7. Tenure Is Strongly Related to Churn Behavior

Churned customers generally have much lower tenure than customers who did not churn.

The tenure distribution and boxplot show that churn is concentrated more heavily among customers with shorter tenure.

This suggests that early-stage customers may represent an important retention segment.

## 8. Higher Monthly Charges Are Associated with Higher Churn

Average monthly charges:

- Churn: 74.44

Median monthly charges:

- Churn: 79.65

Customers who churn show higher monthly charges on average.

This suggests that pricing or service-cost-related factors may be relevant to churn prediction.

## 9. Total Charges Should Be Interpreted Carefully

Average total charges:

- Churn: 1531.80

Median total charges:

- Churn: 703.55

Starting with Phase 3: Feature Engineering and ML pipleling and training testing model on different algos.

```python

# STEP 11 — FEATURE ENGINEERING
# Create a copy so the original cleaned dataframe remains unchanged
df_ml = df.copy()

# 1. Remove identifier / leakage columns

df_ml = df_ml.drop(columns=["customerID", "StateNextMonth"])

# 2. Create target variable
df_ml["State"] = df_ml["State"].map({
    "No Churn": 0,
    "Churn": 1
})

# 3. Tenure-based feature

df_ml["TenureGroup"] = pd.cut(
    df_ml["tenure"],
    bins=[-1, 12, 24, 48, 60, float("inf")],
    labels=["0-1 Year", "1-2 Years", "2-4 Years", "4-5 Years", "5+ Years"]
)

# 4. Average monthly spending

df_ml["AvgMonthlyCharges"] = (
    df_ml["TotalCharges"] / df_ml["tenure"].replace(0, np.nan)
)

df_ml["AvgMonthlyCharges"] = df_ml["AvgMonthlyCharges"].fillna(
    df_ml["MonthlyCharges"]
)

# 5. Service usage feature

service_columns = [
    "PhoneService",
    "MultipleLines",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies"
]

df_ml["ServiceCount"] = (
    df_ml[service_columns]
    .apply(lambda row: sum(value in ["Yes", "Yes, Yes"] for value in row), axis=1)
)

# 6. Security & support feature

security_support_columns = [
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport"
]

df_ml["SecuritySupportCount"] = (
    df_ml[security_support_columns]
    .apply(lambda row: sum(value == "Yes" for value in row), axis=1)
)

# 7. Display the resulting dataset

print("Dataset shape after feature engineering:", df_ml.shape)

print("\nColumns:")
print(df_ml.columns.tolist())

print("\nFirst 5 rows:")
display(df_ml.head())
```

```python

# STEP 12 — TRAIN / TEST SPLIT

from sklearn.model_selection import train_test_split

# Separate features and target
X = df_ml.drop(columns=["State"])
y = df_ml["State"]

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# Check the shapes
print("Original dataset:", X.shape)

print("\nTraining set:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting set:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

# Check target distribution
print("\nTraining target distribution:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True))
```

```python
# STEP 13 — PREPROCESSING PIPELINE

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# Identify numerical and categorical columns
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
# Numerical preprocessing

numeric_transformer = Pipeline(
    steps=[
        ("scaler", StandardScaler())
    ]
)
```

```python
# Categorical preprocessing

categorical_transformer = Pipeline(
    steps=[
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)
```

```python
# Combine preprocessing

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)
```

```python
# Fit ONLY on training data

X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)
```

```python
print("\nOriginal X_train shape:", X_train.shape)
print("Processed X_train shape:", X_train_processed.shape)

print("\nOriginal X_test shape:", X_test.shape)
print("Processed X_test shape:", X_test_processed.shape)
```
